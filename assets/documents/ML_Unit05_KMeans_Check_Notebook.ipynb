```python
# @title
import datetime
from zoneinfo import ZoneInfo
import numpy as np
import sklearn
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

NAME = "Anne Dayer"
# Colab's clock runs on UTC, so ask for Geneva time explicitly
now = datetime.datetime.now(ZoneInfo("Europe/Zurich"))
print("Run by", NAME, "on", now.strftime("%d %B %Y %H:%M %Z"),
      "with scikit-learn", sklearn.__version__)
```

```python
def check(name, X):
    print()
    print(name)
    print(" k   SSE    silhouette")
    for k in range(1, 9):
        # n_init=10: restart k-means 10 times and keep the best run
        km = KMeans(n_clusters=k, n_init=10, random_state=0).fit(X)
        # silhouette needs at least two clusters
        sil = silhouette_score(X, km.labels_) if k > 1 else float("nan")
        print(f"{k:2d}  {km.inertia_:5.1f}   {sil:.2f}")
```

```python
# @title
for seed in (0, 1, 2):
    rng = np.random.default_rng(seed)

    # 250 points spread evenly over a square: no real groups
    uniform = rng.uniform(0, 1, size=(250, 2))
    check(f"Uniform points, sample {seed}", uniform)

    # 4 tight groups of 62 points around the corners of the same square
    corners = [(0.25, 0.25), (0.25, 0.75), (0.75, 0.25), (0.75, 0.75)]
    groups = np.vstack([rng.normal(c, 0.06, size=(62, 2)) for c in corners])
    check(f"Four real groups, sample {seed}", groups)
```
